# Final Evaluation and Takeaways

## Model performance

We evaluated a baseline linear regression model using walk-forward validation
with daily updates (step = 24 hours).

## Key observations

- Hourly electricity demand shows strong daily and weekly seasonality
- Lag-based features capture a significant portion of temporal structure
- Even a simple linear model achieves reasonable predictive performance

## Limitations and next steps

- No hyperparameter tuning was performed
- Only linear models were explored
- Future improvements could include tree-based models or probabilistic forecasting
